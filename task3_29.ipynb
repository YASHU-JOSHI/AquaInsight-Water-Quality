```python
from pathlib import Path
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import ExtraTreesRegressor, RandomForestClassifier, ExtraTreesClassifier
from sklearn.metrics import (
    mean_absolute_error, mean_squared_error, r2_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report,
    accuracy_score, precision_score, recall_score, f1_score,
    precision_recall_curve, average_precision_score
)

RANDOM_STATE = 42
DATA_PATH = Path("Water Quality.csv")
TARGET_NAME = "Specific conductance"

print("Working directory:", Path.cwd())
print("Dataset expected at:", DATA_PATH.resolve())
```

```python
# If the CSV is elsewhere, change DATA_PATH to its full path.
if not DATA_PATH.exists():
    candidates = list(Path.cwd().glob("*.csv"))
    if len(candidates) == 1:
        DATA_PATH = candidates[0]
        print("Using detected CSV:", DATA_PATH)
    else:
        raise FileNotFoundError(
            f"Cannot find {DATA_PATH}. Put Water Quality(2).csv beside this notebook "
            "or set DATA_PATH to the full CSV path."
        )

raw = pd.read_csv(DATA_PATH, low_memory=False)
print("Raw shape:", raw.shape)
display(raw.head())
print("\nColumns:")
print(raw.columns.tolist())
```

```python
required = {"CharacteristicName", "ResultValue"}
missing = required - set(raw.columns)
if missing:
    raise ValueError(f"Missing required columns for this dataset format: {missing}")

df = raw.copy()
df["CharacteristicName"] = df["CharacteristicName"].astype(str).str.strip()
df["ResultValue"] = pd.to_numeric(df["ResultValue"], errors="coerce")

# Build a sample key from available location/date fields.
key_candidates = ["MonitoringLocationID", "ActivityStartDate"]
keys = [c for c in key_candidates if c in df.columns]
if not keys:
    raise ValueError("Expected MonitoringLocationID and/or ActivityStartDate to identify samples.")

if "ActivityStartDate" in df.columns:
    df["ActivityStartDate"] = pd.to_datetime(df["ActivityStartDate"], errors="coerce")

# Use location + date when both exist; otherwise use available key(s).
df = df.dropna(subset=["ResultValue"])
if "ActivityStartDate" in keys:
    df = df.dropna(subset=["ActivityStartDate"])

# Keep numeric measurement parameters. Exclude IDs, dates, and metadata from the pivot.
# Each parameter becomes a feature column; repeated readings are averaged.
wide = df.pivot_table(
    index=keys,
    columns="CharacteristicName",
    values="ResultValue",
    aggfunc="mean"
).reset_index()
wide.columns.name = None

print("Pivoted sample table shape:", wide.shape)
display(wide.head())
```

```python
# Normalize names for robust matching, but preserve original column names.
def find_column_by_phrase(columns, phrase):
    phrase = phrase.casefold()
    matches = [c for c in columns if phrase in str(c).casefold()]
    return matches[0] if matches else None

target_col = find_column_by_phrase(wide.columns, TARGET_NAME)
if target_col is None:
    print("Could not find exact phrase 'Specific conductance'. Available measurement columns include:")
    print([c for c in wide.columns if c not in keys][:50])
    raise ValueError(
        "Specific conductance target not found. Update TARGET_NAME to a numeric characteristic "
        "that exists in your dataset."
    )

# Avoid obvious target leakage from conductivity aliases and the target itself.
metadata_cols = set(keys)
leakage_terms = ("conductivity", "specific conductance")
feature_cols = [
    c for c in wide.columns
    if c not in metadata_cols
    and c != target_col
    and not any(term in str(c).casefold() for term in leakage_terms)
    and pd.api.types.is_numeric_dtype(wide[c])
]

# Require enough observations for training/evaluation.
model_df = wide[[*keys, target_col, *feature_cols]].copy()
model_df = model_df.replace([np.inf, -np.inf], np.nan)
model_df = model_df.dropna(subset=[target_col])
# Remove all-empty and constant features.
feature_cols = [
    c for c in feature_cols
    if model_df[c].notna().any() and model_df[c].nunique(dropna=True) > 1
]
if len(feature_cols) == 0:
    raise ValueError("No usable numeric predictor columns remain after leakage filtering.")

X = model_df[feature_cols].copy()
y = pd.to_numeric(model_df[target_col], errors="coerce")
valid = y.notna()
X, y, model_df = X.loc[valid], y.loc[valid], model_df.loc[valid]

print("Target column:", target_col)
print("Rows:", len(X), "| Numeric predictors:", len(feature_cols))
print("Target summary:")
display(y.describe())
print("Feature columns:")
print(feature_cols)
```

```python
# If date is present, use a chronological holdout; otherwise use a seeded random split.
if "ActivityStartDate" in model_df.columns:
    order = model_df["ActivityStartDate"].sort_values().index
    cut = int(len(order) * 0.80)
    train_idx, test_idx = order[:cut], order[cut:]
    # Ensure no empty split for small datasets.
    if len(train_idx) == 0 or len(test_idx) == 0:
        train_idx, test_idx = train_test_split(
            model_df.index, test_size=0.20, random_state=RANDOM_STATE
        )
else:
    train_idx, test_idx = train_test_split(
        model_df.index, test_size=0.20, random_state=RANDOM_STATE
    )

X_train, X_test = X.loc[train_idx], X.loc[test_idx]
y_train, y_test = y.loc[train_idx], y.loc[test_idx]

print("Train rows:", len(X_train), "| Test rows:", len(X_test))
```

```python
classifier = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", ExtraTreesClassifier(
        n_estimators=300, min_samples_leaf=2,
        class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1
    ))
])
classifier.fit(X_train, y_train_class)
y_pred_class = classifier.predict(X_test)
y_score_class = classifier.predict_proba(X_test)[:, 1]

metrics = {
    "Accuracy": accuracy_score(y_test_class, y_pred_class),
    "Precision": precision_score(y_test_class, y_pred_class, zero_division=0),
    "Recall": recall_score(y_test_class, y_pred_class, zero_division=0),
    "F1": f1_score(y_test_class, y_pred_class, zero_division=0)
}
print("Task 2 metrics:")
display(pd.DataFrame([metrics]).round(4))
print("\nClassification report:")
print(classification_report(
    y_test_class, y_pred_class,
    labels=[0, 1],
    target_names=["At/below median", "Above median"],
    zero_division=0
))
```

```python
# Threshold is learned from training targets only to avoid test-set threshold leakage.
class_threshold = float(y_train.median())
y_train_class = (y_train > class_threshold).astype(int)
y_test_class = (y_test > class_threshold).astype(int)

print(f"Training-set median threshold: {class_threshold:.6g}")
print("Class labels: 0 = at/below training median; 1 = above training median")
print("Test class counts:")
display(y_test_class.value_counts().rename(index={0: "At/below median", 1: "Above median"}))
```

```python
cm = confusion_matrix(y_test_class, y_pred_class, labels=[0, 1])
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["At/below median", "Above median"]
)
disp.plot(values_format="d")
plt.title("Task 2: Confusion matrix")
plt.tight_layout()
plt.show()
```

```python
# Use a validation split from training data to choose the decision threshold.
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train, y_train_class,
    test_size=0.20, random_state=RANDOM_STATE, stratify=y_train_class
)

threshold_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", ExtraTreesClassifier(
        n_estimators=300, min_samples_leaf=2,
        class_weight="balanced", random_state=RANDOM_STATE, n_jobs=-1
    ))
])
threshold_model.fit(X_fit, y_fit)
val_scores = threshold_model.predict_proba(X_val)[:, 1]

prec_val, rec_val, thresholds_val = precision_recall_curve(y_val, val_scores)
if len(thresholds_val):
    f1_vals = (2 * prec_val[:-1] * rec_val[:-1]) / (prec_val[:-1] + rec_val[:-1] + 1e-12)
    best_idx = int(np.nanargmax(f1_vals))
    best_threshold = float(thresholds_val[best_idx])
    best_val_f1 = float(f1_vals[best_idx])
else:
    best_threshold, best_val_f1 = 0.5, float("nan")

print(f"Selected threshold from training validation split: {best_threshold:.4f}")
print(f"Validation F1 at selected threshold: {best_val_f1:.4f}")
```

```python
# Evaluate the already-fitted main classifier on the held-out test set.
test_ap = average_precision_score(y_test_class, y_score_class)
prec_test, rec_test, thresholds_test = precision_recall_curve(y_test_class, y_score_class)

plt.figure(figsize=(7, 6))
plt.plot(rec_test, prec_test, label=f"Extra Trees (AP = {test_ap:.3f})")
baseline_precision = float(y_test_class.mean())
plt.axhline(baseline_precision, linestyle="--", label=f"Positive prevalence = {baseline_precision:.3f}")
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Task 3: Precision–Recall curve (held-out test set)")
plt.legend()
plt.tight_layout()
plt.show()

print(f"Test Average Precision (AUC-PR-style summary): {test_ap:.4f}")
```

```python
# Apply threshold selected on the training validation split, then report held-out test metrics.
y_pred_tuned = (y_score_class >= best_threshold).astype(int)
threshold_metrics = {
    "Threshold": best_threshold,
    "Accuracy": accuracy_score(y_test_class, y_pred_tuned),
    "Precision": precision_score(y_test_class, y_pred_tuned, zero_division=0),
    "Recall": recall_score(y_test_class, y_pred_tuned, zero_division=0),
    "F1": f1_score(y_test_class, y_pred_tuned, zero_division=0),
    "Average Precision (AUC-PR)": test_ap
}
display(pd.DataFrame([threshold_metrics]).round(4))

cm_tuned = confusion_matrix(y_test_class, y_pred_tuned, labels=[0, 1])
ConfusionMatrixDisplay(
    confusion_matrix=cm_tuned,
    display_labels=["At/below median", "Above median"]
).plot(values_format="d")
plt.title("Task 3: Confusion matrix after validation-based threshold selection")
plt.tight_layout()
plt.show()
```
